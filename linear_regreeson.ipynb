```python
df = pd.get_dummies(
    df,
    columns=['ocean_proximity'],
    drop_first=True,
    dtype=int
)
```

```python
df.head(10)
```

```python
x = df.drop(columns=['median_house_value'])
y = df['median_house_value']
```

```python
x.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.shape,x_test.shape
```

```python
y_train.shape,y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression,Lasso,Ridge
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error,root_mean_squared_error
```

```python
lr= LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_predict = lr.predict(x_test)
```

```python
mse = mean_squared_error(y_test,y_predict)
mae = mean_absolute_error(y_test,y_predict)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,y_predict)
```

```python
print(mse)
print(mae)
print(rmse)
print(r2)
```

```python
lass = Lasso(alpha=0.7)
lass.fit(x_train,y_train)
```

```python
lass_predict = lass.predict(x_test)
```

```python
lass_mse = mean_squared_error(y_test,lass_predict)
lass_mae = mean_absolute_error(y_test,lass_predict)
lass_rmse = np.sqrt(lass_mse)
```

```python
print(lass_mse)
print(lass_mae)
print(lass_rmse)
```

```python
lass.score(x_test,y_test)
```

```python
lass.intercept_
```

```python
ridge = Ridge(alpha=1.0)
ridge.fit(x_train,y_train)
```

```python
ridge_predict = ridge.predict(x_test)
```

```python
ridge_mse = mean_squared_error(y_test,ridge_predict)
ridge_mae = mean_absolute_error(y_test,ridge_predict)
ridge_rmse = np.sqrt(ridge_mse)
ridge_r2 = r2_score(y_test,ridge_predict)
```

```python
print(ridge_mse)
print(ridge_mae)
print(ridge_rmse)
print(ridge_r2)
```

```python
ridge.score(x_test,y_test)
```

```python
ridge.intercept_
```

```python
coefs = pd.DataFrame({
    'Linear Regression': a,
    'Lasso Regression': b,
    'Ridge Regression': c
})
```

```python
metrics = pd.DataFrame({
    'Linear Regression': [mse, mae, rmse, r2],
    'Lasso Regression': [lass_mse, lass_mae, lass_rmse,lass.score(x_test,y_test)],
    'Ridge Regression': [ridge_mse, ridge_mae, ridge_rmse,ridge_r2]
}, index=['Mean Squared Error', 'Mean Absolute Error', 'Root Mean Squared Error', 'R-squared'])
```

```python
metrics
```

Ridge Regression is the best among the three, but only by a very small amount.
Your R² values are:
- Linear: 62.996%
- Lasso: 62.997%
- Ridge: 63.001%
So practically, all three models are performing almost identically.
Why is Ridge slightly better?
Ridge adds L2 regularization, which penalizes very large coefficients and can help when features are correlated.
Lasso uses L1 regularization, which can shrink some coefficients toward zero.
Because your features probably don't have a severe overfitting problem, regularization only provides a small improvement.
